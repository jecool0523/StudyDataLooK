```python
import os
import json
import matplotlib.font_manager as fm
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'Do Hyeon'

for dir in os.listdir('./datas'):
    with open(f'./datas/{dir}', 'r', encoding='utf-8') as f:
        print(dir)
        dt = json.load(f)
    sumd = {label: 0.0 for label in labels}
    for gul in dt:
        for label in labels:
            sumd[label] += gul.get(label, 0.0)
    print(sumd)
    sumd = [v for k, v in sumd.items() if v > 0]

    
    cotob = {'white': 'dark', 'black': 'light'}
    for cat_color in ['white', 'black']:
        school_name = dir.split('.')[0]

        cmap = plt.cm.get_cmap('Set3', 11)
        colors = [cmap(i) for i in range(11)]
        labels = ["여성/가족", "남성", "성소수자", "인종/국적", "연령", "지역", "종교", "장애", "기타 혐오", "악플/욕설", "개인지칭"]

        plt.figure(figsize=(6, 6))

        wedges, texts, autotexts = plt.pie(sumd, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
        plt.setp(texts + autotexts, color=cat_color)
        plt.title(f"{school_name}", pad=20, color = cat_color)
        plt.axis('equal')
        plt.tight_layout()

        cat_dir = f'./plots/{cotob[cat_color]}'
        os.makedirs(cat_dir, exist_ok=True)

        save_path = os.path.join(cat_dir, f"{school_name}.png")
        plt.savefig(save_path, dpi=300, transparent=True)

        plt.close()
```
